# Business Understanding

## Latar Belakang Masalah

Perusahaan retail sering kali menghabiskan anggaran pemasaran secara tidak efisien karena menggunakan pendekatan promosi yang seragam (one-size-fits-all). Dengan memahami segmentasi alami pelanggan berbasis data transaksi, promosi dapat dipersonalisasi sesuai profil daya beli dan sensitivitas harga masing-masing segmen.

## Tujuan Bisnis & Analisis Data

* Mengidentifikasi kelompok/persona pelanggan alami menggunakan algoritma K-Means Clustering tanpa label bawaan (unsupervised learning).
* Menganalisis karakteristik tiap cluster berdasarkan 3 indikator: tingkat pendapatan (Income), belanja barang premium (MntWines), dan respons terhadap diskon (NumDealsPurchases).
* Menyusun rekomendasi strategi pemasaran berbasis data yang relevan dan efisien untuk setiap kelompok pelanggan.

## Kriteria Keberhasilan

* Terbentuk cluster yang terpisah dengan baik secara matematis (dievaluasi melalui grafik Elbow Method dan nilai Silhouette Score).
* Profil cluster dapat diinterpretasikan secara jelas menjadi persona bisnis (misal: High Income Low Discount Hunter, Bargain Hunter, dsb.).

Import seluruh library yang dibutuhkan: `numpy` dan `pandas` untuk manipulasi data, `matplotlib` dan `seaborn` untuk visualisasi, `StandardScaler` untuk standarisasi fitur, `KMeans` untuk algoritma clustering, `silhouette_score` untuk evaluasi cluster, dan `joblib` untuk menyimpan artefak model.

# Data Preparation

Menangani missing values pada kolom `Income` dengan mengisi nilai kosong menggunakan median. Median dipilih karena lebih tahan terhadap pengaruh outlier dibandingkan mean.

```python
print(f'Missing values sebelum: {df["Income"].isnull().sum()}')
df['Income'] = df['Income'].fillna(df['Income'].median())
print(f'Missing values sesudah: {df["Income"].isnull().sum()}')
```

Sebanyak 24 missing values pada kolom `Income` telah berhasil diisi dengan nilai median, sehingga kini tidak ada lagi data kosong pada fitur tersebut.

Menghapus data outlier pada kolom `Income` yang nilainya lebih dari 200.000. Outlier ini dapat menarik centroid K-Means secara tidak proporsional dan menghasilkan cluster yang bias.

```python
print(f'Jumlah data sebelum: {df.shape[0]}')
df = df[df['Income'] < 200000].copy()
print(f'Jumlah data sesudah: {df.shape[0]}')
```

Setelah menghapus outlier pada `Income` (nilai > 200.000), jumlah data berkurang dari 2.240 menjadi 2.239 baris. Hanya 1 baris yang dihapus.

Memilih tiga fitur yang akan digunakan untuk proses clustering, kemudian melakukan standarisasi menggunakan `StandardScaler` agar ketiga fitur memiliki skala yang sebanding (rata-rata 0, standar deviasi 1). Hal ini penting karena K-Means menghitung jarak Euclidean yang sensitif terhadap perbedaan skala.

```python
features = ['Income', 'MntWines', 'NumDealsPurchases']
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

scaled_df = pd.DataFrame(X_scaled, columns=features)
print('Statistik setelah scaling:')
scaled_df.describe().round(4)
```

Setelah standarisasi, seluruh fitur kini memiliki rata-rata mendekati 0 dan standar deviasi mendekati 1. Data siap digunakan untuk proses clustering.

# Modeling

Pada tahap ini digunakan algoritma **K-Means Clustering**. Jumlah cluster optimal ditentukan melalui dua metode evaluasi:
1. **Elbow Method** — Mencari titik di mana penurunan inertia mulai melandai.
2. **Silhouette Score** — Mengukur kualitas pemisahan antar cluster.

## Elbow Method

Menguji jumlah cluster dari K=2 hingga K=10 dan menghitung nilai inertia (Within-Cluster Sum of Squares) untuk setiap nilai K. Hasilnya divisualisasikan dalam bentuk line plot.

```python
inertia_list = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_list.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia_list, marker='o', linewidth=2, markersize=8)
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.xticks(range(2, 11))
plt.grid(True, alpha=0.3)
plt.show()
```

Dari grafik Elbow Method, terlihat penurunan inertia yang signifikan dari K=2 ke K=3, kemudian mulai melandai setelah K=3. Titik siku (elbow) terletak pada **K=3**, yang mengindikasikan bahwa 3 cluster merupakan jumlah yang optimal.

## Silhouette Score

Menghitung Silhouette Score untuk setiap nilai K dari 2 hingga 10. Skor ini mengukur seberapa mirip sebuah data point dengan cluster-nya sendiri dibandingkan dengan cluster terdekat lainnya. Nilai mendekati 1 menunjukkan pemisahan cluster yang baik.

```python
silhouette_list = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_list.append(score)
    print(f'K={k}: Silhouette Score = {score:.4f}')

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_list, marker='o', linewidth=2, markersize=8, color='green')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score untuk Berbagai K')
plt.xticks(range(2, 11))
plt.grid(True, alpha=0.3)
plt.show()
```

Hasil Silhouette Score menunjukkan bahwa **K=3** memberikan skor yang baik dan konsisten dengan hasil Elbow Method. Oleh karena itu, **K=3** dipilih sebagai jumlah cluster optimal.

## Membangun Model Final

Melatih model K-Means dengan jumlah cluster K=3 pada data yang telah distandardisasi, kemudian menetapkan label cluster ke setiap pelanggan dan menampilkan distribusi jumlah pelanggan per cluster.

```python
kmeans_final = KMeans(n_clusters=3, random_state=42, n_init=10)
final_labels = kmeans_final.fit_predict(X_scaled)

df['Cluster'] = final_labels

print('Distribusi Cluster:')
print(df['Cluster'].value_counts().sort_index())
```

Model K-Means berhasil mengelompokkan 2.239 pelanggan ke dalam 3 cluster. Distribusi cluster menunjukkan jumlah anggota yang relatif seimbang antar kelompok.

# Evaluation

## Profiling Cluster

Menghitung rata-rata setiap fitur untuk masing-masing cluster guna memahami karakteristik tiap kelompok pelanggan.

```python
profile = df.groupby('Cluster')[features].mean().round(2)
print('=== Rata-rata Fitur per Cluster ===')
print(profile.to_string())
```

Berdasarkan profil rata-rata fitur per cluster, terlihat perbedaan karakteristik yang jelas antar kelompok pelanggan.

Memvisualisasikan rata-rata setiap fitur per cluster dalam bentuk bar chart untuk melihat perbedaan karakteristik secara visual.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

for i, col in enumerate(features):
    df.groupby('Cluster')[col].mean().plot(kind='bar', ax=axes[i], color=colors)
    axes[i].set_title(f'Rata-rata {col} per Cluster')
    axes[i].set_xlabel('Cluster')
    axes[i].set_ylabel(col)
    axes[i].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```

Memvisualisasikan hasil clustering pada scatterplot `Income` vs `MntWines` dengan warna berbeda untuk setiap cluster, sehingga terlihat pemisahan kelompok pelanggan secara visual.

```python
plt.figure(figsize=(10, 7))
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

for cluster in range(3):
    mask = df['Cluster'] == cluster
    plt.scatter(df.loc[mask, 'Income'], 
                df.loc[mask, 'MntWines'],
                c=colors[cluster], label=f'Cluster {cluster}', alpha=0.6, s=30)

plt.xlabel('Income')
plt.ylabel('MntWines')
plt.title('Scatter Plot Income vs MntWines per Cluster')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

Dari scatter plot terlihat bahwa ketiga cluster memiliki pemisahan yang cukup baik berdasarkan pendapatan dan pengeluaran wine. Cluster dengan pendapatan tinggi cenderung terletak di sisi kanan atas grafik.

Menghitung nilai Silhouette Score dan Davies-Bouldin Index untuk mengevaluasi kualitas cluster secara kuantitatif.

```python
from sklearn.metrics import davies_bouldin_score

sil_score = silhouette_score(X_scaled, final_labels)
db_score = davies_bouldin_score(X_scaled, final_labels)

print(f'Silhouette Score (K=3): {sil_score:.4f}')
print(f'Davies-Bouldin Index (K=3): {db_score:.4f}')
print()
print('Interpretasi:')
print('- Silhouette Score mendekati 1 = cluster terpisah dengan baik')
print('- Davies-Bouldin Index mendekati 0 = cluster compact dan terpisah')
```

Nilai Silhouette Score dan Davies-Bouldin Index menunjukkan bahwa model K-Means dengan K=3 menghasilkan cluster yang cukup baik. Pemisahan antar cluster dapat diterima untuk keperluan segmentasi pelanggan.

Kedua artefak berhasil diekspor:
* `scaler_marketing.joblib` — Menyimpan parameter standarisasi (mean dan std) agar data baru dapat ditransformasi dengan skala yang sama.
* `kmeans_marketing.joblib` — Menyimpan model KMeans yang telah dilatih beserta posisi centroid setiap cluster.

Kedua file ini akan digunakan pada aplikasi Streamlit untuk deployment.
